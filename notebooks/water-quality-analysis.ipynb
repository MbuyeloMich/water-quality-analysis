```python
print("🔍 CHECKING MISSING VALUES IN HEAVY METALS")
print("=" * 50)

def check_metal_completeness(df, metals_list, dataset_name):
    print(f"\n{dataset_name} - Metal Data Completeness:")
    print("-" * 40)
    
    results = []
    for metal in metals_list:
        if metal in df.columns:
            non_zero = (df[metal] > 0).sum()
            total = len(df[metal].dropna())
            if total > 0:
                percentage = (non_zero / total) * 100
                results.append((metal, non_zero, total, percentage))
    
    # Sort by detection percentage
    results.sort(key=lambda x: x[3], reverse=True)
    
    for metal, non_zero, total, percentage in results:
        status = "✅" if percentage > 50 else "⚠️" if percentage > 10 else "❌"
        print(f"{status} {metal:<6}: {non_zero:>3}/{total:<3} detected ({percentage:>5.1f}%)")
    
    return results

# Check both datasets
print("\n" + "="*50)
mhlathuze_results = check_metal_completeness(mhlathuze_df, mhlathuze_metals, "MHLATHUZE")

print("\n" + "="*50)
luvuvu_results = check_metal_completeness(luvuvu_df, luvuvu_metals, "LUVUVU")
```

```python
print("🔗 CHECKING CORRELATIONS WITH pH")
print("=" * 50)

# Function to check correlation of metals with pH
def check_ph_correlations(df, dataset_name):
    print(f"\n{dataset_name} - Correlation with pH")
    print("-" * 40)
    
    if 'pH' not in df.columns:
        print("No pH data available")
        return
    
    correlations = []
    
    # Check each numeric column (excluding pH itself)
    for col in df.select_dtypes(include=[np.number]).columns:
        if col != 'pH' and df[col].notna().sum() > 10:  # Need enough data
            correlation = df['pH'].corr(df[col])
            if not np.isnan(correlation):
                correlations.append((col, correlation))
    
    # Sort by absolute correlation strength
    correlations.sort(key=lambda x: abs(x[1]), reverse=True)
    
    # Print top 10 correlations
    print(f"{'Parameter':<15} {'Correlation with pH':>20}")
    print("-" * 40)
    
    for param, corr in correlations[:10]:  # Top 10 only
        strength = "STRONG" if abs(corr) > 0.5 else "MODERATE" if abs(corr) > 0.3 else "WEAK"
        direction = "positive" if corr > 0 else "negative"
        print(f"{param:<15} {corr:>7.3f} ({strength} {direction})")
    
    return correlations

# Check both datasets
mhlathuze_corrs = check_ph_correlations(mhlathuze_df, "MHLATHUZE")
print("\n" + "="*50)
luvuvu_corrs = check_ph_correlations(luvuvu_df, "LUVUVU")
```

# 🔥 MAJOR DISCOVERY: pH Correlations Are OPPOSITE!

## 🎯 KEY FINDING:
- **MHLATHUZE**: All parameters show WEAK NEGATIVE correlation with pH
  - As pH increases, metals/ions slightly DECREASE
  - Most correlations around -0.2 (very weak)

- **LUVUVU**: Strong POSITIVE correlations with pH!
  - Boron (B): 0.607 correlation - VERY STRONG!
  - Sodium (Na): 0.602 - VERY STRONG!
  - Sulfate (SO4): 0.596 - VERY STRONG!
  - 7 parameters show STRONG correlations (>0.5)

## 🧪 SCIENTIFIC INTERPRETATION:
1. **Different Water Chemistry**: These are fundamentally different water systems
2. **LUVUVU has Geogenic Influence**: Strong positive correlations suggest mineral dissolution increases with pH
3. **MHLATHUZE might have Anthropogenic Pollution**: Weak negative correlations could indicate external contamination
4. **Boron is Key in LUVUVU**: B=0.607! Boron often comes from:
   - Natural weathering of rocks
   - Agricultural runoff (fertilizers)
   - Industrial wastewater

```python
print("🧪 QUICK TEST: Calculate Ion Balance")
print("=" * 50)

def calculate_ion_balance(df, dataset_name):
    print(f"\n{dataset_name} - Ion Balance Check")
    print("-" * 40)
    
    # Convert concentrations to meq/L (milliequivalents per liter)
    # Cations: Ca, Mg, Na, K
    # Anions: Cl, SO4, HCO3, NO3
    
    cations = 0
    anions = 0
    
    # Cations (divide by equivalent weight)
    if 'Ca' in df.columns:
        cations += df['Ca'] / 20.04  # Ca²⁺: 40.08 g/mol ÷ 2 = 20.04
    if 'Mg' in df.columns:
        cations += df['Mg'] / 12.15  # Mg²⁺: 24.31 ÷ 2 = 12.15
    if 'Na' in df.columns:
        cations += df['Na'] / 22.99  # Na⁺: 22.99 ÷ 1 = 22.99
    if 'K' in df.columns:
        cations += df['K'] / 39.10   # K⁺: 39.10 ÷ 1 = 39.10
    
    # Anions
    if 'Cl' in df.columns:
        anions += df['Cl'] / 35.45   # Cl⁻: 35.45 ÷ 1 = 35.45
    if 'SO4' in df.columns:
        anions += df['SO4'] / 48.03  # SO4²⁻: 96.06 ÷ 2 = 48.03
    if 'HCO3' in df.columns:
        anions += df['HCO3'] / 61.02 # HCO3⁻: 61.02 ÷ 1 = 61.02
    if 'NO3' in df.columns:
        anions += df['NO3'] / 62.00  # NO3⁻: 62.00 ÷ 1 = 62.00
    
    # Calculate balance error
    balance_error = ((cations - anions) / (cations + anions)) * 100
    
    print(f"Samples analyzed: {len(df)}")
    print(f"Mean cation sum: {cations.mean():.2f} meq/L")
    print(f"Mean anion sum: {anions.mean():.2f} meq/L")
    print(f"Mean balance error: {balance_error.mean():.1f}%")
    
    # Check quality
    good_quality = (abs(balance_error) < 10).sum()
    percentage = (good_quality / len(df)) * 100
    print(f"Samples with good ion balance (<10% error): {good_quality}/{len(df)} ({percentage:.1f}%)")
    
    return balance_error

# Calculate for both datasets
print("\n" + "="*50)
mhlathuze_balance = calculate_ion_balance(mhlathuze_df, "MHLATHUZE")

print("\n" + "="*50)
luvuvu_balance = calculate_ion_balance(luvuvu_df, "LUVUVU")
```

# 💧 Hypothesis 4: Water Source Mixing

## The Theory:
Different water sources mixing can create complex chemistry:
1. **Acidic rainwater** + **Alkaline groundwater** = Complex pH patterns
2. **Surface water** + **Groundwater** = Different chemical signatures
3. **Different flow paths** = Different mineral contact times

## What We'll Test:
1. **Total Dissolved Solids (TDS)** patterns
2. **Ion correlation networks**
3. **Statistical clustering** of water types
4. **Mixing model** calculations

```python
print("🧪 TESTING HYPOTHESIS 4: Water Source Mixing")
print("=" * 60)

print("\n📊 METHOD: TDS and Ion Correlation Analysis")
print("-" * 40)

def analyze_water_mixing(df, dataset_name):
    print(f"\n{dataset_name} - Water Mixing Indicators")
    print("-" * 30)
    
    mixing = {}
    
    # 1. TDS variability (indicates different sources)
    if 'TDS' in df.columns:
        tds_cv = (df['TDS'].std() / df['TDS'].mean()) * 100
        mixing['TDS_CV'] = tds_cv
        print(f"TDS Coefficient of Variation: {tds_cv:.1f}%")
        print(f"  >50% = Likely multiple water sources")
        print(f"  20-50% = Moderate mixing")
        print(f"  <20% = Homogeneous source")
    
    # 2. Check if TDS correlates with major ions (indicating mineral dissolution)
    if 'TDS' in df.columns:
        major_ions = ['Ca', 'Mg', 'Na', 'K', 'Cl', 'SO4', 'HCO3']
        available_ions = [ion for ion in major_ions if ion in df.columns]
        
        print(f"\nCorrelation of TDS with major ions:")
        for ion in available_ions:
            corr = df['TDS'].corr(df[ion])
            if not np.isnan(corr):
                strength = "Strong" if abs(corr) > 0.7 else "Moderate" if abs(corr) > 0.4 else "Weak"
                print(f"  {ion}: {corr:.3f} ({strength})")
                mixing[f'TDS_{ion}_corr'] = corr
    
    # 3. Calculate charge balance error variability (indicates mixing)
    if 'Ca' in df.columns and 'Mg' in df.columns and 'Na' in df.columns and 'K' in df.columns:
        # Simple charge balance (cations vs anions)
        cations = df['Ca']/20.04 + df['Mg']/12.15 + df['Na']/22.99 + df['K']/39.10
        anions = 0
        if 'Cl' in df.columns:
            anions += df['Cl']/35.45
        if 'SO4' in df.columns:
            anions += df['SO4']/48.03
        if 'HCO3' in df.columns:
            anions += df['HCO3']/61.02
            
        if anions.sum() > 0:
            charge_error = abs((cations - anions) / (cations + anions)) * 100
            charge_cv = (charge_error.std() / charge_error.mean()) * 100
            mixing['Charge_CV'] = charge_cv
            print(f"\nCharge balance error variability: {charge_cv:.1f}%")
            print(f"  >100% = Very different water sources mixing")
            print(f"  50-100% = Moderate mixing")
            print(f"  <50% = Similar water sources")
    
    return mixing

# Analyze both datasets
print("\n" + "="*60)
mhlathuze_mixing = analyze_water_mixing(mhlathuze_df, "MHLATHUZE")

print("\n" + "="*60)
luvuvu_mixing = analyze_water_mixing(luvuvu_df, "LUVUVU")
```
